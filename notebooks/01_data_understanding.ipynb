```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load original dataset
file_path = #insert your file path here as r"file_path"

df = pd.read_csv(file_path)

# Initial inspection
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```
# Identify monthly observations

month_mask = df["c1"].astype(str).str.match(
    r"^[A-Za-z]{3}-\d{4}$"
)

monthly_data = df.loc[month_mask].copy()

monthly_data["Date"] = pd.to_datetime(
    monthly_data["c1"],
    format="%b-%Y"
)

print("Total monthly observations:",
      len(monthly_data))

print("First observation:",
      monthly_data["Date"].min())

print("Last observation:",
      monthly_data["Date"].max())

print("Duplicate months:",
      monthly_data["Date"].duplicated().sum())

print("Missing values:",
      monthly_data.isnull().sum().sum())
```

```
# Extract the core payment indicators

payment_data = monthly_data[
    ["Date", "c30", "c31", "c34", "c35", "c36", "c37"]
].copy()

# Rename columns

payment_data.columns = [
    "Date",
    "IMPS_Volume_Lakh",
    "IMPS_Value_Crore",
    "NEFT_Volume_Lakh",
    "NEFT_Value_Crore",
    "UPI_Volume_Lakh",
    "UPI_Value_Crore"
]

# Convert numeric columns

numeric_columns = payment_data.columns.drop("Date")

for col in numeric_columns:
    payment_data[col] = pd.to_numeric(
        payment_data[col],
        errors="coerce"
    )

# Arrange dates chronologically

payment_data = payment_data.sort_values(
    "Date"
).reset_index(drop=True)

# Inspect the cleaned dataset

print("Dataset Shape:", payment_data.shape)

print("\nFirst 5 Observations:")
display(payment_data.head())

print("\nLast 5 Observations:")
display(payment_data.tail())

print("\nData Types:")
print(payment_data.dtypes)

print("\nMissing Values:")
print(payment_data.isnull().sum())

print("\nDescriptive Statistics:")
display(payment_data.describe())

# Save processed dataset

payment_data.to_csv(
    #insert your file path here as r"file_path",
    index=False
)
```
